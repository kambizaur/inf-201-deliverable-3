### 1. Classes

#### 1.1. Character
The `Character` class represents a basic character in the RPG game. It defines the common attributes and behaviours shared by all characters, such as health points (HP), damage, weapons and the ability to attack, take damage and heal. It also provides properties to access the character's current health and check whether the character is still alive. This class serves as the base class for more specialized character types, such as `Warrior` and `Mage`.

Attributes
- `name (str)`: Stores the character's name.
- `_hp (int)`: Stores the character's current health points. It is initialized with the same value as `hp_max`. It is treated as a private attribute to discourage direct access or modification.
- `hp_max (int)`: Defines the maximum health points the character can have. If no value is specified, the default value is 100.
- `damage (int)`: Defines the base damage dealt when the character attacks without using a weapon. If no value is specified, the default value is 10.
- `weapon (Weapon | None)`: Stores a Weapon object associated with the character. If no weapon is provided, its value is None.

Properties (read-only)
- `hp (int)`: Returns the character's current HP without allowing direct assignment.
- `is_alive (bool)`: Returns True if the character's current HP is greater than zero, and False otherwise.

Methods
- `__init__(name, hp_max, damage, weapon)`: The constructor that initializes a new character object with a name, maximum HP, base damage and an optional weapon.
- `take_damage(damage_points)`: Reduces the character's current HP by the specified damage points. HP cannot go below zero.
- `heal(heal_points)`: Restores the character's HP by the specified amount, without exceeding the maximum HP.
- `attack(target)`: Attacks another character. If the character has a weapon, the damage is determined by rolling the weapon's damage dice. Otherwise, the character uses its base damage.
- `introduce()`: Prints a short introduction containing the character's name.
- `__repr__() -> str`: Returns a representation of the character object, including its name, HP and damage.
- `__str__() -> str`: Returns a sentence describing the character, including its name, current HP and damage.

#### 1.2. Warrior
The `Warrior` class represents a specialized character with defensive abilities. It inherits the attributes and methods from the `Character` base class and introduces an additional armor attribute. The warrior takes reduced damage when attacked, as the armor value is subtracted from the incoming damage.

Attributes
- `armor (int)`: Defines the warrior's armor level, which reduces the damage received from attacks. If no value is specified, the default value is 5.

Attributes (inherited from the Character class)
- `name (str)`
- `_hp (int)`
- `hp_max (int)`
- `damage (int)`
- `weapon (Weapon | None)`

Methods
- `__init__(name, hp_max, damage, weapon, armor)`: Initializes a warrior by calling the constructor of the `Character` base class using super(). It also initializes the warrior's armor attribute. If no value is specified, the armor has a default value of 5.
- `take_damage(damage_points)`: Overrides the `take_damage()` method from the `Character` class. It reduces incoming damage by the warrior's armor value before applying it to the character's HP. The damage reduction cannot result in negative damage.

#### 1.3. Mage
The `Mage` class represents a specialized character with magical abilities. It inherits the attributes and methods from the `Character` base class and introduces a mana system that allows the mage to perform magical attacks. The mage can use magic attacks when sufficient mana is available. If mana is insufficient, the mage performs a regular attack and regenerates some mana.

Attributes
- `mana (int)`: Represents the mage's current mana, which is consumed when performing magical attacks. If no value is specified, the default is 30.
- `mana_max (int)`: Defines the maximum amount of mana the mage can have. If no value is specified, the default is 30.
- `SPELL_COST (int)`: Defines the mana cost of a magical attack. It has a constant value of 10.
- `SPELL_MULTIPLIER (int)`: Defines the damage multiplier applied to magical attacks. It has a constant value of 2.
- `MANA_REGEN (int)`: Defines the amount of mana restored when the mage performs a regular attack due to insufficient mana. It has a constant value of 5.

Attributes (inherited from the Character class)
- `name (str)`
- `_hp (int)`
- `hp_max (int)`
- `damage (int)`
- `weapon (Weapon | None)`

Methods
- `__init__(name, hp_max, damage, weapon, mana, mana_max)`: Initializes a mage by calling the constructor of the `Character` base class using super(). It also initializes the mage's current and maximum mana. The default values for both mana attributes are 30.
- `attack(target)`: Overrides the `attack()` method from the `Character` class. If the mage has enough mana (at least 10), it performs a magical attack, dealing twice its base damage and consuming 10 mana points. Otherwise, it performs a regular attack using the inherited `attack()` method and restores 5 mana points, without exceeding the maximum mana.

#### 1.4. Weapon

The Weapon class represents a weapon that a character can. It stores the weapon's name and the dice used to determine the damage dealt during an attack. Each weapon is associated with a Dice object, which is responsible for generating the damage value.

Attributes
- `name (str)`: Stores the name of the weapon.
- `damage_dice (Dice)`: Stores a Dice object used to determine the damage dealt by the weapon when an attack is performed.

Methods
- `__init__(name, damage_dice)`: The constructor that initializes a new weapon object with its name and associated damage dice.
- `roll_damage() -> int`: Rolls the weapon's damage dice and returns the resulting damage value as an integer.
- `__repr__() -> str`: Returns a string representation of the weapon object, displaying its name and associated damage dice.



#### 1.5. Dice
The Dice class represents a n-sided dice. It is responsible for generating random numbers within a specified range, from 1 to the number of sides.

Attributes
- `side(int)`: Store the number of sides of the dice. The value must be at least 2. If no value is specified during initialization, it defaults to 6.

Methods
- `__init__(sides)`: The constructor that initializes a new dice object with specific number of sides. It raises a ValueError if the number of sides is less than 2.
- `roll()`: Generates and returns a random integer between 1 and the number of sides, using Python's `random.randint()` function.
- `__repr__() -> str`: Returns a string representation of the dice object, displaying its number of sides.
- `__call__() -> int`: Allows the dice object to be called like a regular function. When called, it performs a dice roll and returns the resulting value by invoking the `roll()` method.

#### 1.6. Battle

The Battle class manages the combat between two characters (a hero and an enemy). It controls the battle rounds, ensuring that each character gets an opportunity to attack according to the game rules. The battle continues until either the hero or the enemy is defeated.

Attributes
- `hero (Character)`: Stores the Character object representing the hero participating in the battle.
- `enemy (Character)`: Stores the Character object representing the enemy participating in the battle.

Methods
- `__init__(hero, enemy)`: The constructor initializes a battle by assigning the hero and the enemy Character objects to their respective attributes.

- `play_round()`: Executes one round of combat. First, the hero attacks the enemy. If the enemy is still alive after the attack, it gets the opportunity to attack the hero.

- `simulate_battle()`: Runs the entire battle by repeatedly calling play_round(). The loop continues as long as both characters are alive. It stops when either character is defeated.

#### 2.0. Imports

```python
import sys
from pathlib import Path

sys.path.append(str(Path("..").resolve() / "src"))

from battle import Battle
from character import Mage, Warrior
from dice import Dice
from item import Weapon
import random
```

#### 2.1. Weapons

```python
dagger = Weapon("Dagger", Dice(10))
sword = Weapon("Sword", Dice(20))
```

#### 2.2. Characters

```python
gandalf = Mage(
    name="Gandalf the Grey",
    hp_max=100,
    damage=7,
    mana_max=100,
    mana=100
)

saruman = Mage(
    name="Saruman the White",
    hp_max=100,
    damage=8,
    weapon=dagger,
    mana_max=100,
    mana=100
)

aragorn = Warrior(
    name="Aragorn",
    hp_max=120,
    damage=8,
    weapon=sword,
    armor=5
)

ugluk = Warrior(
    name="Uglúk",
    hp_max=130,
    damage=9,
    armor=3
)
```

#### 2.3. Battles

```python
# Permutate to randomize the character order.
characters = [gandalf, saruman, aragorn, ugluk]
random.shuffle(characters)
for index_1, char_1 in enumerate(characters):
    for char_2 in characters[index_1+1:]:
        # Roll dice to decide who takes first turn.
        first_turn = Dice(2).roll() 

        if first_turn == 1:
            battle = Battle(char_1, char_2)
            print(f"Battle between {char_1.name} and {char_2.name}!")
        else:
            battle = Battle(char_2, char_1)
            print(f"Battle between {char_2.name} and {char_1.name}!")

        battle.simulate_battle()

        if char_1.is_alive:
            print(f"The winner is {char_1.name}!")
        else:
            print(f"The winner is {char_2.name}!")
        print(f"\n")

        # We heal them to prepare for next battle
        char_1.heal(char_1.hp_max)
        char_2.heal(char_2.hp_max)
```
