#### 2.0. Imports

```python
import sys
from pathlib import Path

sys.path.append(str(Path("..").resolve() / "src"))

from battle import Battle
from character import Mage, Warrior
from dice import Dice
from item import Weapon
from numpy.random import permutation
```

#### 2.1. Weapons

```python
dagger = Weapon("Dagger", Dice(10))
sword = Weapon("Sword", Dice(20))
```

#### 2.2. Characters

```python
gandalf = Mage(
    name="Gandalf the Grey",
    hp_max=100,
    damage=7,
    mana_max=100,
    mana=100
)

saruman = Mage(
    name="Saruman the White",
    hp_max=100,
    damage=8,
    weapon=dagger,
    mana_max=100,
    mana=100
)

aragorn = Warrior(
    name="Aragorn",
    hp_max=120,
    damage=8,
    weapon=sword,
    armor=5
)

ugluk = Warrior(
    name="Uglúk",
    hp_max=130,
    damage=9,
    armor=3
)
```

#### 2.3. Battles

```python
# Permutate to randomize the character order.
characters = permutation([gandalf, saruman, aragorn, ugluk])
for index_1, char_1 in enumerate(characters):
    for char_2 in characters[index_1+1:]:
        # Roll dice to decide who takes first turn.
        first_turn = Dice(2).roll() 

        if first_turn == 1:
            battle = Battle(char_1, char_2)
            print(f"Battle between {char_1.name} and {char_2.name}!")
        else:
            battle = Battle(char_2, char_1)
            print(f"Battle between {char_2.name} and {char_1.name}!")

        battle.simulate_battle()

        if char_1.is_alive:
            print(f"The winner is {char_1.name}!")
        else:
            print(f"The winner is {char_2.name}!")
        print(f"\n")

        # We heal them to prepare for next battle
        char_1.heal(char_1.hp_max)
        char_2.heal(char_2.hp_max)
```
